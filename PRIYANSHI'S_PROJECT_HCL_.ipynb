```python
import pandas as pd

df = pd.read_csv('/wholesale customers (2).csv')

df.head()
```

```python
# Count customers in each channel
print("Customers in Each Channel:")
print(df['Channel'].value_counts())

# Count customers in each region
print("\nCustomers in Each Region:")
print(df['Region'].value_counts())

# Channel and Region comparison
print("\nChannel vs Region:")
print(pd.crosstab(df['Channel'], df['Region']))
```

```python
# Select product categories
categories = ['Fresh', 'Milk', 'Grocery', 'Frozen',
              'Detergents_Paper', 'Delicassen']

# Calculate total spending
total_spending = df[categories].sum()

print("Total Spending by Category:")
print(total_spending.sort_values(ascending=False))

# Calculate average spending
average_spending = df[categories].mean()

print("\nAverage Spending by Category:")
print(average_spending.sort_values(ascending=False))

# Highest total spending
print("\nHighest Total Spending Category:")
print(total_spending.idxmax())

# Highest average spending
print("\nHighest Average Spending Category:")
print(average_spending.idxmax())
```

```python
# NumPy statistical calculations
import numpy as np
for col in categories:


    print("\nCategory:", col)

    print("Mean:", np.mean(df[col]))
    print("Median:", np.median(df[col]))
    print("Minimum:", np.min(df[col]))
    print("Maximum:", np.max(df[col]))
```

```python
import numpy as np

# Select product categories
categories = ['Fresh', 'Milk', 'Grocery', 'Frozen',
              'Detergents_Paper', 'Delicassen']

# Calculate total spending of each customer using NumPy
total_spending = np.sum(df[categories].values, axis=1)

# Find indices of top 10 customers
top_10_indices = np.argsort(total_spending)[-10:][::-1]

# Display top 10 customers
top_10 = df.iloc[top_10_indices].copy()

top_10['Total_Spending'] = total_spending[top_10_indices]

print("Top 10 Customers with Highest Spending:")
print(top_10[['Channel', 'Region', 'Total_Spending']])
```

```python
# Group data by channel and calculate average spending

channel_comparison = df.groupby('Channel')[categories].mean()

print("Average Category Spending by Channel:")
print(channel_comparison)
```

```python
# Group data by channel and calculate average spending

channel_comparison = df.groupby('Channel')[categories].mean()

print("Average Category Spending by Channel:")
print(channel_comparison)
```

```python
# Observation 1: Most popular channel

most_popular_channel = df['Channel'].value_counts().idxmax()

print("Observation 1:")
print("Channel", most_popular_channel,
      "has the highest number of customers.")


# Observation 2: Highest spending category

categories = ['Fresh', 'Milk', 'Grocery', 'Frozen',
              'Detergents_Paper', 'Delicassen']

total_spending = df[categories].sum()

highest_category = total_spending.idxmax()

print("\nObservation 2:")
print(highest_category,
      "has the highest total customer spending.")


# Observation 3: Highest average spending category

average_spending = df[categories].mean()

highest_average_category = average_spending.idxmax()

print("\nObservation 3:")
print(highest_average_category,
      "has the highest average customer spending.")
```
